```python
INFERENCE_PKG_PATH = (
    "/content/drive/MyDrive/seq2seq_checkpoints/"
    "seq2seq_inference_package.pt"
)

print("File exists:", os.path.exists(INFERENCE_PKG_PATH))

package = torch.load(
    INFERENCE_PKG_PATH,
    map_location=device,
    weights_only=False
)

print("Inference package loaded!")
print("Epoch:", package["trained_epoch"])
print("Validation loss:", package["val_loss"])
print("Config:", package["config"])
```

```python
config = package["config"]

EMB_DIM = config["EMB_DIM"]
HID_DIM = config["HID_DIM"]
NUM_LAYERS = config["NUM_LAYERS"]
DROPOUT = config["DROPOUT"]
MAX_LEN = config["MAX_LEN"]
MAX_DECODE_LEN = config["MAX_DECODE_LEN"]
BEAM_SIZE = config["BEAM_SIZE"]
REVERSE_SOURCE = config["REVERSE_SOURCE"]

src_stoi = package["src_stoi"]
src_itos = package["src_itos"]

tgt_stoi = package["tgt_stoi"]
tgt_itos = package["tgt_itos"]

PAD = package["specials"]["PAD"]
SOS = package["specials"]["SOS"]
EOS = package["specials"]["EOS"]
UNK = package["specials"]["UNK"]

print("Source vocabulary:", len(src_stoi))
print("Target vocabulary:", len(tgt_stoi))
print("Beam size:", BEAM_SIZE)
print("Reverse source:", REVERSE_SOURCE)
```

```python
class Encoder(nn.Module):
    def __init__(self, vocab_size, emb_dim, hid_dim, num_layers, pad_idx, dropout):
        super().__init__()

        self.embedding = nn.Embedding(
            vocab_size,
            emb_dim,
            padding_idx=pad_idx
        )

        self.lstm = nn.LSTM(
            emb_dim,
            hid_dim,
            num_layers,
            batch_first=True,
            dropout=dropout if num_layers > 1 else 0.0
        )

    def forward(self, src, src_lens):
        emb = self.embedding(src)

        packed = nn.utils.rnn.pack_padded_sequence(
            emb,
            src_lens.cpu(),
            batch_first=True,
            enforce_sorted=False
        )

        _, (h, c) = self.lstm(packed)

        return h, c


class Decoder(nn.Module):
    def __init__(self, vocab_size, emb_dim, hid_dim, num_layers, pad_idx, dropout):
        super().__init__()

        self.embedding = nn.Embedding(
            vocab_size,
            emb_dim,
            padding_idx=pad_idx
        )

        self.lstm = nn.LSTM(
            emb_dim,
            hid_dim,
            num_layers,
            batch_first=True,
            dropout=dropout if num_layers > 1 else 0.0
        )

        self.fc_out = nn.Linear(
            hid_dim,
            vocab_size
        )

    def forward(self, tgt_in, h, c):
        emb = self.embedding(tgt_in)

        out, (h, c) = self.lstm(
            emb,
            (h, c)
        )

        logits = self.fc_out(out)

        return logits, h, c


class Seq2Seq(nn.Module):
    def __init__(self, encoder, decoder):
        super().__init__()

        self.encoder = encoder
        self.decoder = decoder

    def forward(self, src, src_lens, tgt_in):
        h, c = self.encoder(src, src_lens)

        logits, _, _ = self.decoder(
            tgt_in,
            h,
            c
        )

        return logits
```

```python
encoder = Encoder(
    len(src_itos),
    EMB_DIM,
    HID_DIM,
    NUM_LAYERS,
    src_stoi[PAD],
    DROPOUT
)

decoder = Decoder(
    len(tgt_itos),
    EMB_DIM,
    HID_DIM,
    NUM_LAYERS,
    tgt_stoi[PAD],
    DROPOUT
)

model = Seq2Seq(
    encoder,
    decoder
).to(device)

model.load_state_dict(package["model_state"])

model.eval()

print("Model loaded successfully!")
```

```python
mt_en = MosesTokenizer(lang="en")
mdt_fr = MosesDetokenizer(lang="fr")
```

```python
@torch.no_grad()
def beam_search_decode(
    model,
    src_tokens,
    beam_size=BEAM_SIZE,
    max_len=MAX_DECODE_LEN
):
    model.eval()

    # Convert English tokens to IDs
    src_ids = [
        src_stoi.get(token, src_stoi[UNK])
        for token in src_tokens
    ]

    # Reverse source because training used REVERSE_SOURCE=True
    if REVERSE_SOURCE:
        src_ids = src_ids[::-1]

    # Add EOS
    src_ids = src_ids + [src_stoi[EOS]]

    src = torch.tensor(
        [src_ids],
        dtype=torch.long,
        device=device
    )

    src_lens = torch.tensor(
        [len(src_ids)],
        dtype=torch.long
    )

    # Encode source sentence
    h0, c0 = model.encoder(
        src,
        src_lens
    )

    # (log probability, token IDs, hidden state, cell state, finished)
    beams = [
        (
            0.0,
            [tgt_stoi[SOS]],
            h0,
            c0,
            False
        )
    ]

    completed = []

    for _ in range(max_len):

        candidates = []

        for log_prob, ids, h, c, finished in beams:

            if finished:
                candidates.append(
                    (
                        log_prob,
                        ids,
                        h,
                        c,
                        True
                    )
                )
                continue

            inp = torch.tensor(
                [[ids[-1]]],
                dtype=torch.long,
                device=device
            )

            logits, h_new, c_new = model.decoder(
                inp,
                h,
                c
            )

            log_probs = F.log_softmax(
                logits[:, -1, :],
                dim=-1
            ).squeeze(0)

            topk_lp, topk_idx = log_probs.topk(
                beam_size
            )

            for lp, idx in zip(
                topk_lp.tolist(),
                topk_idx.tolist()
            ):

                done = (
                    idx == tgt_stoi[EOS]
                )

                candidates.append(
                    (
                        log_prob + lp,
                        ids + [idx],
                        h_new,
                        c_new,
                        done
                    )
                )

        # Keep top beam_size candidates
        candidates.sort(
            key=lambda x: x[0],
            reverse=True
        )

        beams = candidates[:beam_size]

        completed.extend(
            [
                b for b in beams
                if b[4]
            ]
        )

        if all(
            b[4] for b in beams
        ):
            break

    # Select best completed beam
    best = max(
        completed if completed else beams,
        key=lambda x: x[0]
    )

    ids = [
        i
        for i in best[1][1:]
        if i != tgt_stoi[EOS]
    ]

    return [
        tgt_itos[i]
        for i in ids
    ]
```

```python
def translate(sentence):
    # English tokenization
    src_tokens = mt_en.tokenize(
        sentence,
        escape=False
    )

    # Generate French tokens
    output_tokens = beam_search_decode(
        model,
        src_tokens
    )

    # French detokenization
    translation = mdt_fr.detokenize(
        output_tokens
    )

    return translation
```

```python
sentences =[ "The European Union needs a common policy.",
           "This is a very important decision."]


for sentence in sentences:
    print("EN:", sentence)
    print("FR:", translate(sentence))
    print("-" * 80)
```
